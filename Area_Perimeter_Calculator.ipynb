```python
print("WELCOME TO AREA AND PERIMETER CALCULATIONS CALCULATOR")
while True:
    print("What you want to calculate \n1.Area \n2.Perimeter")
    user = input()
    if user == '1':
        print("Which shape's area do you want to calculate:")
        print("1.Square \n2.Rectangle \n3.Circle \n4.Triangle \n5.Parallelogram")
        nm = input()
        if nm == '1':
            try:
                square = float(input("Enter the side's length of square:"))
                square1 = square**2
                print(f"The area of square is {square1}")
            except ValueError:
                print("Please enter a valid number!")
        elif nm == '2':
            try:
                length = float(input("Enter the length:"))
                width = float(input("Enter the width:"))
                rectangle = length*width
                print(f"The area of rectangle is {rectangle}")
            except ValueError:
                print("Please enter a valid number!")
        elif nm == '3':
            try:
                rad = float(input("Enter the length of radius:"))
                radius = 3.14159*(rad**2)
                print(f"The area of circle is {radius}")
            except ValueError:
                print("Please enter a valid number!")
        elif nm == '4':
            try:
                base = float(input('Enter the length of base:'))
                height = float(input('Enter the length of height:'))
                triangle = 0.5*base*height
                print(f"The area of triangle is {triangle}")
            except ValueError:
                print("Please enter a valid number!")
        elif nm == '5':
            try:
                base1 = float(input('Enter the length of base:'))
                height1 = float(input('Enter the length of height:'))
                parallelogram = base1*height1
                print(f"The area of parallelogram is {parallelogram}")
            except ValueError:
                print("Please enter a valid number!")
        else:
            print("Invalid..")
    elif user == '2':
        print("Which shape's perimeter do you want to calculate:")
        print("1.Square \n2.Rectangle \n3.Circle \n4.Triangle \n5.Parallelogram")
        nm = input()
        if nm == '1':
            try:
                square = float(input("Enter the side's length of square:"))
                square1 = 4*square
                print(f"The perimeter of square is {square1}")
            except ValueError:
                print("Please enter a valid number!")
        elif nm == '2':
            try:
                length = float(input("Enter the length:"))
                width = float(input("Enter the width:"))
                rectangle = 2*(length+width)
                print(f"The perimeter of rectangle is {rectangle}")
            except ValueError:
                print("Please enter a valid number!")
        elif nm == '3':
            try:
                rad = float(input("Enter the length of radius:"))
                radius = 2*3.14159*rad
                print(f"The perimeter of circle is {radius}")
            except ValueError:
                print("Please enter a valid number!")
        elif nm == '4':
            try:
                a = float(input('Enter the length of side 1:'))
                b = float(input('Enter the length of side 2:'))
                c = float(input('Enter the length of side 3:'))
                triangle = a+b+c
                print(f"The perimeter of triangle is {triangle}")
            except ValueError:
                print("Please enter a valid number!")
        elif nm == '5':
            try:
                base1 = float(input('Enter the length of base:'))
                height1 = float(input('Enter the length of height:'))
                parallelogram = 2*(base1+height1)
                print(f"The perimeter of parallelogram is {parallelogram}")
            except ValueError:
                print("Please enter a valid number!")
        else:
            print("Invalid..")
    else:
        print("Invalid...")
    print("Do you want to continue y/n")
    y = input()
    if y == 'y':
        continue
    elif y == 'n':
        break
    else:
        print("Invalid input!")
print("Thank you for using Area & Perimeter Calculator!")
```
